## Data Setup

#### Import the dataset from the UCI ML Repository using the provided steps.

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
ozone_level_detection = fetch_ucirepo(id=172) 
  
# data (as pandas dataframes) 
X = ozone_level_detection.data.features 
y = ozone_level_detection.data.targets 
  
# metadata 
print(ozone_level_detection.metadata) 
  
# variable information 
print(ozone_level_detection.variables) 
```

```python
import pandas as pd
import numpy as np

ids = ozone_level_detection.data.ids
df = pd.concat([ids, X, y], axis=1)
df.shape
```

```python
df["Dataset"].value_counts()
```

#### Save data to the github repository

```python
df.to_csv("../data/ozone.csv", index=False)

eighthr = df[df["Dataset"] == "8hr"].copy()
onehr = df[df["Dataset"] == "1hr"].copy()
```

## Data Inspection

```python
#check columns
print('8hr Data:', eighthr.select_dtypes(exclude="number").columns.tolist())

print('1hr Data:', onehr.select_dtypes(exclude="number").columns.tolist())
```

```python
#check for missing values
print('8hr Data:')
print("Percent of rows with missing features:", eighthr.drop(columns=["Dataset", "Date"]).isna().any(axis=1).mean())
print("Count of '?':", (eighthr == "?").sum().sum())

print('\n1hr Data:')
print("Percent of rows with missing features:", onehr.drop(columns=["Dataset", "Date"]).isna().any(axis=1).mean())
print("Count of '?':", (onehr == "?").sum().sum())
```

```python
#parse dates
print('8hr Data:')
eighthr["Date"] = pd.to_datetime(eighthr["Date"])
eighthr = eighthr.sort_values("Date").reset_index(drop=True)
print(eighthr["Date"].head())

print('\n1hr Data:')
onehr["Date"] = pd.to_datetime(onehr["Date"])
onehr = onehr.sort_values("Date").reset_index(drop=True)
print(onehr["Date"].head())
```

```python
#inspect date information
print('8hr Data:')
print("Date range:", eighthr["Date"].min(), "to", eighthr["Date"].max())
print("Duplicate dates:", eighthr["Date"].duplicated().sum())
print("Missing calendar days:", len(pd.date_range(eighthr["Date"].min(), eighthr["Date"].max())) - len(eighthr))

print('\n1hr Data:')
print("Date range:", onehr["Date"].min(), "to", onehr["Date"].max())
print("Duplicate dates:", onehr["Date"].duplicated().sum())
print("Missing calendar days:", len(pd.date_range(onehr["Date"].min(), onehr["Date"].max())) - len(onehr))
```

```python
#check targets
print('8hr Data:')
print(eighthr["Class"].value_counts())
print(eighthr["Class"].mean())

print('\n1hr Data:')
print(onehr["Class"].value_counts())
print(onehr["Class"].mean())
```

```python
#ensure features are numeric
eighthr.drop(columns=["Dataset", "Date", "Class"]).select_dtypes(exclude="number").columns
onehr.drop(columns=["Dataset", "Date", "Class"]).select_dtypes(exclude="number").columns
```

#### Save cleaned-up files

```python
eighthr.to_csv("../data/ozone_eighthr.csv", index=False)
onehr.to_csv("../data/ozone_onehr.csv", index=False)
```
